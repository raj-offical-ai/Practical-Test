```python
# ============================================================
# SMART EXPENSE TRACKER - FINAL JUPYTER NOTEBOOK VERSION
# ============================================================

from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# CONSTANTS
# ============================================================

DATA_FILE = Path("expenses.csv")

DATE_FORMAT = "%Y-%m-%d"

COLUMNS = [
    "Date",
    "Amount",
    "Category",
    "Description"
]

VALID_CATEGORIES = [
    "Food",
    "Transport",
    "Utilities",
    "Entertainment",
    "Shopping",
    "Healthcare",
    "Education",
    "Other"
]


# ============================================================
# CREATE FRESH CSV
# ============================================================

def create_fresh_csv():

    empty_data = pd.DataFrame(
        columns=COLUMNS
    )

    empty_data.to_csv(
        DATA_FILE,
        index=False
    )


# ============================================================
# EXPENSE TRACKER CLASS
# ============================================================

class ExpenseTracker:

    def __init__(self, filename=DATA_FILE):

        self.filename = Path(filename)

        self.df = self._load_data()


    # ========================================================
    # LOAD DATA
    # ========================================================

    def _load_data(self):

        if not self.filename.exists():

            empty_data = pd.DataFrame(
                columns=COLUMNS
            )

            empty_data.to_csv(
                self.filename,
                index=False
            )

            return empty_data


        try:

            df = pd.read_csv(
                self.filename
            )

        except Exception:

            print(
                "Old CSV file could not be read."
            )

            print(
                "Creating a fresh CSV file..."
            )

            empty_data = pd.DataFrame(
                columns=COLUMNS
            )

            empty_data.to_csv(
                self.filename,
                index=False
            )

            return empty_data


        # Check CSV columns
        if list(df.columns) != COLUMNS:

            print(
                "CSV format is incorrect."
            )

            print(
                "Creating a fresh CSV file..."
            )

            empty_data = pd.DataFrame(
                columns=COLUMNS
            )

            empty_data.to_csv(
                self.filename,
                index=False
            )

            return empty_data


        # Convert Date
        df["Date"] = pd.to_datetime(
            df["Date"],
            errors="coerce"
        )


        # Convert Amount
        df["Amount"] = pd.to_numeric(
            df["Amount"],
            errors="coerce"
        )


        # Clean Category
        df["Category"] = (
            df["Category"]
            .astype(str)
            .str.strip()
            .str.title()
        )


        # Clean Description
        df["Description"] = (
            df["Description"]
            .astype(str)
            .str.strip()
        )


        # Remove invalid records
        df = df.dropna(
            subset=[
                "Date",
                "Amount"
            ]
        )


        # Amount must be positive
        df = df[
            df["Amount"] > 0
        ]


        # Keep valid categories
        df = df[
            df["Category"].isin(
                VALID_CATEGORIES
            )
        ]


        # Sort by date
        df = (
            df
            .sort_values("Date")
            .reset_index(drop=True)
        )


        return df


    # ========================================================
    # SAVE DATA
    # ========================================================

    def _save_data(self):

        output = self.df.copy()


        if not output.empty:

            output["Date"] = (
                output["Date"]
                .dt.strftime(
                    DATE_FORMAT
                )
            )


        output.to_csv(
            self.filename,
            index=False
        )


    # ========================================================
    # ADD EXPENSE
    # ========================================================

    def add_expense(
        self,
        date,
        amount,
        category,
        description
    ):

        # ----------------------------------------------------
        # DATE VALIDATION
        # ----------------------------------------------------

        try:

            parsed_date = datetime.strptime(
                date.strip(),
                DATE_FORMAT
            )

        except:

            print(
                "\nInvalid date!"
            )

            print(
                "Please enter date like: 2026-09-10"
            )

            return False


        # ----------------------------------------------------
        # AMOUNT VALIDATION
        # ----------------------------------------------------

        try:

            amount = float(
                amount
            )

        except:

            print(
                "\nAmount must be numeric!"
            )

            return False


        if amount <= 0:

            print(
                "\nAmount must be greater than zero!"
            )

            return False


        # ----------------------------------------------------
        # CATEGORY VALIDATION
        # ----------------------------------------------------

        category = (
            category
            .strip()
            .title()
        )


        if category not in VALID_CATEGORIES:

            print(
                "\nInvalid category!"
            )

            print(
                "Available categories:"
            )

            print(
                ", ".join(
                    VALID_CATEGORIES
                )
            )

            return False


        # ----------------------------------------------------
        # DESCRIPTION VALIDATION
        # ----------------------------------------------------

        description = (
            description
            .strip()
        )


        if description == "":

            print(
                "\nDescription cannot be empty!"
            )

            return False


        # ----------------------------------------------------
        # CREATE DATA
        # ----------------------------------------------------

        new_expense = pd.DataFrame([
            {
                "Date":
                    pd.Timestamp(
                        parsed_date
                    ),

                "Amount":
                    amount,

                "Category":
                    category,

                "Description":
                    description
            }
        ])


        # Add new row
        self.df = pd.concat(
            [
                self.df,
                new_expense
            ],
            ignore_index=True
        )


        # Sort data
        self.df = (
            self.df
            .sort_values("Date")
            .reset_index(drop=True)
        )


        # Save data
        self._save_data()


        print(
            "\nExpense added successfully!"
        )


        return True


    # ========================================================
    # SUMMARY
    # ========================================================

    def get_summary(self):

        if self.df.empty:

            return {
                "total": 0,
                "average": 0,
                "count": 0,
                "highest": 0,
                "lowest": 0,
                "category_totals":
                    pd.Series(dtype=float)
            }


        # NumPy Array
        amounts = self.df[
            "Amount"
        ].to_numpy(
            dtype=float
        )


        # NumPy calculations
        total = np.sum(
            amounts
        )

        average = np.mean(
            amounts
        )

        highest = np.max(
            amounts
        )

        lowest = np.min(
            amounts
        )

        count = len(
            amounts
        )


        # Pandas calculation
        category_totals = (
            self.df
            .groupby(
                "Category"
            )["Amount"]
            .sum()
            .sort_values(
                ascending=False
            )
        )


        return {

            "total":
                float(total),

            "average":
                float(average),

            "count":
                count,

            "highest":
                float(highest),

            "lowest":
                float(lowest),

            "category_totals":
                category_totals
        }


    # ========================================================
    # MONTHLY METRICS
    # ========================================================

    def monthly_metrics(self):

        if self.df.empty:

            return pd.DataFrame(
                columns=[
                    "Month",
                    "Total",
                    "Average"
                ]
            )


        temp = self.df.copy()


        temp["Month"] = (
            temp["Date"]
            .dt.to_period("M")
            .astype(str)
        )


        result = (
            temp
            .groupby("Month")["Amount"]
            .agg(
                Total="sum",
                Average="mean"
            )
            .reset_index()
        )


        return result


    # ========================================================
    # SPENDING PERCENTAGE
    # ========================================================

    def spending_percentage_by_category(
        self
    ):

        if self.df.empty:

            return pd.DataFrame(
                columns=[
                    "Category",
                    "Amount",
                    "Percentage"
                ]
            )


        result = (
            self.df
            .groupby("Category")
            .agg(
                Amount=("Amount", "sum")
            )
            .reset_index()
        )


        total = np.sum(
            result["Amount"]
            .to_numpy()
        )


        if total > 0:

            result["Percentage"] = (
                result["Amount"]
                / total
                * 100
            )

        else:

            result["Percentage"] = 0


        return (
            result
            .sort_values(
                "Amount",
                ascending=False
            )
            .reset_index(drop=True)
        )


    # ========================================================
    # FILTER EXPENSES
    # ========================================================

    def filter_expenses(
        self,
        category=None,
        start_date=None,
        end_date=None,
        min_amount=None,
        max_amount=None
    ):

        filtered = self.df.copy()


        # Category
        if category:

            category = (
                category
                .strip()
                .title()
            )

            filtered = filtered[
                filtered["Category"]
                == category
            ]


        # Start date
        if start_date:

            try:

                start = pd.Timestamp(
                    datetime.strptime(
                        start_date,
                        DATE_FORMAT
                    )
                )

                filtered = filtered[
                    filtered["Date"]
                    >= start
                ]

            except:

                print(
                    "Invalid start date!"
                )

                return pd.DataFrame()


        # End date
        if end_date:

            try:

                end = pd.Timestamp(
                    datetime.strptime(
                        end_date,
                        DATE_FORMAT
                    )
                )

                filtered = filtered[
                    filtered["Date"]
                    <= end
                ]

            except:

                print(
                    "Invalid end date!"
                )

                return pd.DataFrame()


        # Minimum amount
        if min_amount is not None:

            try:

                filtered = filtered[
                    filtered["Amount"]
                    >= float(
                        min_amount
                    )
                ]

            except:

                print(
                    "Invalid minimum amount!"
                )

                return pd.DataFrame()


        # Maximum amount
        if max_amount is not None:

            try:

                filtered = filtered[
                    filtered["Amount"]
                    <= float(
                        max_amount
                    )
                ]

            except:

                print(
                    "Invalid maximum amount!"
                )

                return pd.DataFrame()


        return (
            filtered
            .reset_index(drop=True)
        )


    # ========================================================
    # SUMMARY REPORT
    # ========================================================

    def generate_report(self):

        summary = self.get_summary()


        print("\n")
        print("=" * 65)
        print(
            "          SMART EXPENSE TRACKER REPORT"
        )
        print("=" * 65)


        print(
            f"Number of expenses : "
            f"{summary['count']}"
        )


        print(
            f"Total spending     : "
            f"₹{summary['total']:,.2f}"
        )


        print(
            f"Average expense    : "
            f"₹{summary['average']:,.2f}"
        )


        print(
            f"Highest expense    : "
            f"₹{summary['highest']:,.2f}"
        )


        print(
            f"Lowest expense     : "
            f"₹{summary['lowest']:,.2f}"
        )


        if summary["count"] == 0:

            print(
                "\nNo expenses available."
            )

            print("=" * 65)

            return


        # ----------------------------------------------------
        # CATEGORY
        # ----------------------------------------------------

        print(
            "\nCategory-wise Spending:"
        )

        print("-" * 40)


        for category, amount in (
            summary[
                "category_totals"
            ].items()
        ):

            print(
                f"{category:<20}"
                f"₹{amount:,.2f}"
            )


        # ----------------------------------------------------
        # MONTHLY
        # ----------------------------------------------------

        print(
            "\nMonthly Spending:"
        )


        monthly = (
            self.monthly_metrics()
        )


        print(
            monthly.to_string(
                index=False,
                formatters={
                    "Total":
                        "₹{:,.2f}".format,

                    "Average":
                        "₹{:,.2f}".format
                }
            )
        )


        # ----------------------------------------------------
        # PERCENTAGE
        # ----------------------------------------------------

        print(
            "\nSpending Percentage by Category:"
        )


        percentage = (
            self
            .spending_percentage_by_category()
        )


        print(
            percentage.to_string(
                index=False,
                formatters={
                    "Amount":
                        "₹{:,.2f}".format,

                    "Percentage":
                        "{:.2f}%".format
                }
            )
        )


        print("=" * 65)


    # ========================================================
    # FINAL FIXED VISUALIZATION
    # ========================================================

    def visualize(self):

        if self.df.empty:

            print(
                "\nNo data available for visualization."
            )

            print(
                "Please add some expenses first."
            )

            return


        # ----------------------------------------------------
        # SEABORN STYLE
        # ----------------------------------------------------

        sns.set_theme(
            style="whitegrid"
        )


        # ----------------------------------------------------
        # PREPARE DATA
        # ----------------------------------------------------

        df = self.df.copy()

        df["Date"] = pd.to_datetime(
            df["Date"]
        )


        # Category-wise spending
        category_data = (
            df
            .groupby("Category")["Amount"]
            .sum()
            .sort_values(
                ascending=False
            )
            .reset_index()
        )


        # Monthly spending
        monthly = (
            df
            .assign(
                Month=df["Date"]
                .dt.to_period("M")
                .astype(str)
            )
            .groupby("Month")["Amount"]
            .sum()
            .reset_index()
        )


        # Daily spending
        daily = (
            df
            .groupby("Date")["Amount"]
            .sum()
            .reset_index()
            .sort_values("Date")
        )


        # ----------------------------------------------------
        # CREATE LARGE DASHBOARD
        # ----------------------------------------------------

        fig, axes = plt.subplots(
            2,
            2,
            figsize=(20,16)
        )


        fig.suptitle(
            "SMART EXPENSE TRACKER - VISUALIZATION DASHBOARD",
            fontsize=20,
            fontweight="bold",
            y=0.98
        )


        # ====================================================
        # 1. BAR CHART
        # ====================================================

        sns.barplot(
            data=category_data,
            x="Category",
            y="Amount",
            ax=axes[0, 0]
        )


        axes[0, 0].set_title(
            "Category-wise Spending",
            fontsize=15,
            fontweight="bold",
            pad=12
        )


        axes[0, 0].set_xlabel(
            "Category",
            fontsize=11
        )


        axes[0, 0].set_ylabel(
            "Amount (₹)",
            fontsize=11
        )


        axes[0, 0].tick_params(
            axis="x",
            rotation=35
        )


        # Add values on bars
        for container in axes[0, 0].containers:

            axes[0, 0].bar_label(
                container,
                fmt="₹%.0f",
                padding=3,
                fontsize=9
            )


        # ====================================================
        # 2. LINE CHART - FIXED
        # ====================================================

        if len(daily) >= 2:

            axes[0, 1].plot(
                daily["Date"],
                daily["Amount"],
                marker="o",
                linewidth=2.5,
                markersize=7
            )


            axes[0, 1].set_title(
                "Daily Spending Trend",
                fontsize=15,
                fontweight="bold",
                pad=12
            )


            axes[0, 1].set_xlabel(
                "Date",
                fontsize=11
            )


            axes[0, 1].set_ylabel(
                "Spending (₹)",
                fontsize=11
            )


            # Show every date clearly
            axes[0, 1].set_xticks(
                daily["Date"]
            )


            axes[0, 1].set_xticklabels(
                daily["Date"].dt.strftime("%d-%b"),
                rotation=45,
                ha="right"
            )


            # Add values to points
            for x, y in zip(
                daily["Date"],
                daily["Amount"]
            ):

                axes[0, 1].annotate(
                    f"₹{y:,.0f}",
                    (x, y),
                    textcoords="offset points",
                    xytext=(0, 8),
                    ha="center",
                    fontsize=8
                )


            axes[0, 1].grid(
                True,
                linestyle="--",
                alpha=0.4
            )


        else:

            axes[0, 1].bar(
                daily["Date"].dt.strftime("%d-%b"),
                daily["Amount"]
            )

            axes[0, 1].set_title(
                "Daily Spending Trend",
                fontsize=15,
                fontweight="bold"
            )

            axes[0, 1].set_xlabel(
                "Date"
            )

            axes[0, 1].set_ylabel(
                "Spending (₹)"
            )


        # ====================================================
        # 3. PIE CHART
        # ====================================================

        axes[1, 0].pie(
            category_data["Amount"],
            labels=category_data["Category"],
            autopct="%1.1f%%",
            startangle=90,
            textprops={
                "fontsize": 9
            }
        )


        axes[1, 0].set_title(
            "Spending Distribution by Category",
            fontsize=15,
            fontweight="bold",
            pad=12
        )


        # ====================================================
        # 4. HISTOGRAM - FIXED
        # ====================================================

        amounts = df["Amount"].to_numpy(
            dtype=float
        )


        # Automatic bin calculation
        if len(amounts) == 1:

            bins = 1

        elif len(amounts) <= 5:

            bins = len(amounts)

        else:

            bins = min(
                10,
                max(
                    5,
                    int(
                        np.sqrt(
                            len(amounts)
                        )
                    )
                )
            )


        sns.histplot(
            amounts,
            bins=bins,
            kde=False,
            edgecolor="black",
            ax=axes[1, 1]
        )


        axes[1, 1].set_title(
            "Expense Amount Frequency",
            fontsize=15,
            fontweight="bold",
            pad=12
        )


        axes[1, 1].set_xlabel(
            "Expense Amount (₹)",
            fontsize=11
        )


        axes[1, 1].set_ylabel(
            "Frequency",
            fontsize=11
        )


        axes[1, 1].grid(
            axis="y",
            linestyle="--",
            alpha=0.4
        )


        # ====================================================
        # FINAL SPACING
        # ====================================================

        plt.tight_layout(
            rect=[
                0,
                0,
                1,
                0.95
            ],
            pad=3.0,
            h_pad=3.0,
            w_pad=3.0
        )


        plt.show()


# ============================================================
# HELPER FUNCTION
# ============================================================

def get_non_empty(prompt):

    while True:

        try:

            value = input(
                prompt
            ).strip()

        except KeyboardInterrupt:

            print(
                "\nInput cancelled."
            )

            return ""


        if value:

            return value


        print(
            "Input cannot be empty!"
        )


# ============================================================
# ADD EXPENSE MENU
# ============================================================

def add_expense_menu(tracker):

    print("\n")
    print("-" * 60)
    print(
        "                 ADD NEW EXPENSE"
    )
    print("-" * 60)


    print(
        "Available Categories:"
    )


    print(
        ", ".join(
            VALID_CATEGORIES
        )
    )


    date = get_non_empty(
        "Date (YYYY-MM-DD): "
    )

    if not date:
        return


    amount = get_non_empty(
        "Amount: "
    )

    if not amount:
        return


    category = get_non_empty(
        "Category: "
    )

    if not category:
        return


    description = get_non_empty(
        "Description: "
    )

    if not description:
        return


    tracker.add_expense(
        date,
        amount,
        category,
        description
    )


# ============================================================
# SHOW ALL EXPENSES
# ============================================================

def show_all_expenses(tracker):

    if tracker.df.empty:

        print(
            "\nNo expenses found."
        )

        return


    display_data = (
        tracker.df.copy()
    )


    display_data["Date"] = (
        display_data["Date"]
        .dt.strftime(
            DATE_FORMAT
        )
    )


    print("\n")
    print("-" * 70)
    print(
        "                    ALL EXPENSES"
    )
    print("-" * 70)


    print(
        display_data.to_string(
            index=False
        )
    )


    print("-" * 70)


    print(
        f"Total Spending: "
        f"₹{tracker.df['Amount'].sum():,.2f}"
    )


# ============================================================
# FILTER MENU
# ============================================================

def filter_menu(tracker):

    if tracker.df.empty:

        print(
            "\nNo expenses available for filtering."
        )

        return


    print("\n")
    print("-" * 60)
    print(
        "                  FILTER EXPENSES"
    )
    print("-" * 60)


    print(
        "Leave a field blank to skip it."
    )


    category = input(
        "Category: "
    ).strip() or None


    start_date = input(
        "Start date (YYYY-MM-DD): "
    ).strip() or None


    end_date = input(
        "End date (YYYY-MM-DD): "
    ).strip() or None


    min_text = input(
        "Minimum amount: "
    ).strip()


    max_text = input(
        "Maximum amount: "
    ).strip()


    # Validate amounts
    try:

        min_amount = (
            float(min_text)
            if min_text
            else None
        )

        max_amount = (
            float(max_text)
            if max_text
            else None
        )

    except:

        print(
            "\nAmount must be numeric!"
        )

        return


    result = tracker.filter_expenses(
        category=category,
        start_date=start_date,
        end_date=end_date,
        min_amount=min_amount,
        max_amount=max_amount
    )


    if result.empty:

        print(
            "\nNo matching expenses found."
        )

        return


    display_data = (
        result.copy()
    )


    display_data["Date"] = (
        display_data["Date"]
        .dt.strftime(
            DATE_FORMAT
        )
    )


    print(
        "\nMatching Expenses:"
    )


    print(
        display_data.to_string(
            index=False
        )
    )


    print(
        f"\nFiltered Total: "
        f"₹{result['Amount'].sum():,.2f}"
    )


# ============================================================
# MAIN MENU
# ============================================================

def show_menu():

    print("\n")
    print("=" * 65)
    print(
        "                 SMART EXPENSE TRACKER"
    )
    print(
        "                  Personal Finance Analyzer"
    )
    print("=" * 65)


    print(
        "1. Add Expense"
    )

    print(
        "2. View All Expenses"
    )

    print(
        "3. Generate Summary Report"
    )

    print(
        "4. Filter Expenses"
    )

    print(
        "5. Show Visualizations"
    )

    print(
        "6. Exit"
    )

    print("=" * 65)


# ============================================================
# MAIN PROGRAM
# ============================================================

def main():

    # --------------------------------------------------------
    # FRESH START
    # --------------------------------------------------------

    create_fresh_csv()


    # --------------------------------------------------------
    # CREATE TRACKER
    # --------------------------------------------------------

    tracker = ExpenseTracker(
        DATA_FILE
    )


    print(
        "\nSmart Expense Tracker started successfully!"
    )


    # --------------------------------------------------------
    # MENU LOOP
    # --------------------------------------------------------

    while True:

        show_menu()


        try:

            choice = input(
                "Enter your choice (1-6): "
            ).strip()

        except KeyboardInterrupt:

            print(
                "\n\nProgram stopped by user."
            )

            break


        # ====================================================
        # OPTION 1
        # ====================================================

        if choice == "1":

            add_expense_menu(
                tracker
            )


        # ====================================================
        # OPTION 2
        # ====================================================

        elif choice == "2":

            show_all_expenses(
                tracker
            )


        # ====================================================
        # OPTION 3
        # ====================================================

        elif choice == "3":

            tracker.generate_report()


        # ====================================================
        # OPTION 4
        # ====================================================

        elif choice == "4":

            filter_menu(
                tracker
            )


        # ====================================================
        # OPTION 5
        # ====================================================

        elif choice == "5":

            tracker.visualize()


        # ====================================================
        # OPTION 6
        # ====================================================

        elif choice == "6":

            print("\n")
            print("=" * 65)

            print(
                "     Thank you for using Smart Expense Tracker!"
            )

            print(
                "              Program completed."
            )

            print("=" * 65)

            break


        # ====================================================
        # INVALID OPTION
        # ====================================================

        else:

            print(
                "\nInvalid choice!"
            )

            print(
                "Please select a number from 1 to 6."
            )


# ============================================================
# START PROGRAM
# ============================================================

main()
```
